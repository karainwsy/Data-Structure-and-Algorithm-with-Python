##### ***基础排序算法***
###### 排序算法（英语：Sorting algorithm）是一种能将一串资料依照特定排序方式排列的算法，排序后的资料即可放在有序数组。<br>**排序**是指将一组数据（例如一个数组或列表）按照某种特定的顺序（升序或降序）重新排列的过程。<br>排序的依据通常是数据元素的某个**关键码**，比如数字的大小、字符串的字典序等。<br>排序是计算机科学中最基础、最核心的算法问题之一。无论是整理通讯录、分析考试成绩，还是数据库查询优化，排序算法都无处不在。<br>理解不同的排序算法，不仅能帮助我们解决实际问题，更是深入理解算法设计与分析思想的绝佳途径。
###### 排序算法可以从多个维度进行分类，最常见的分类方式是基于其时间复杂度和空间复杂度：

###### 而对于那些时间复杂度较高的算法，我将其称之为基础排序算法，本节主要介绍这些基础排序算法，如：冒泡排序、选择排序、插入排序。


##### ***冒泡排序***
###### 核心思想：重复"遍历"待排序序列，一次比较两个相邻元素，如果它们的顺序错误就把它们交换过来。每一轮遍历都会将当前未排序部分的最大（或最小）元素冒泡到正确位置。<br>**算法步骤**：<br>1. 比较相邻的元素。如果第一个比第二个大（升序），就交换它们两个。<br>2. 对每一对相邻元素做同样的工作，从开始第一对到结尾的最后一对。这步做完后，最后的元素会是最大的数。<br>3. 针对所有的元素重复以上的步骤，除了最后一个（因为每一轮都会确定一个最终元素）。<br>4. 持续每次对越来越少的元素重复上面的步骤，直到没有任何一对数字需要比较。
###### **性能分析**：<br>1. 时间复杂度：最好情况 O(n)（已有序时，可通过优化实现），最坏和平均情况 O(n²)。<br>2. 空间复杂度：O(1)，是原地排序。<br>3. 稳定性：稳定。

```python
# 实现
def bubble_sort(arr):
    n = len(arr)
    for i in range(n-1):
        for j in range(n-1-i):
            if arr[j] > arr[j+1]:
                temp = arr[j]
                arr[j] = arr[j+1]
                arr[j+1] = temp
    return arr

test_data = [64, 34, 25, 12, 22, 11, 90]
print("原始数组:", test_data)
sorted_data = bubble_sort(test_data.copy()) # 使用副本，避免修改原数据
print("冒泡排序后:", sorted_data)
```

##### ***选择排序***
###### 核心思想：在未排序序列中找到最小（或最大）元素，存放到排序序列的起始位置，然后再从剩余未排序元素中继续寻找最小（大）元素，放到已排序序列的末尾。以此类推，直到所有元素均排序完毕。<br>**算法步骤**：<br>1. 初始状态：整个序列为无序区。<br>2. 第 i 轮（i 从 0 开始）排序时，当前无序区为 arr[i...n-1]。从该区中选出关键字最小的记录 arr[min_index]。<br>3. 将 arr[min_index] 与无序区的第一个记录 arr[i] 交换。这样，arr[0...i] 就形成了有序区。<br>4. n-1 轮结束后，数组排序完成。
###### **性能分析**：<br>1. 时间复杂度：始终为 O(n²)，因为无论数据是否有序，都需要进行完整的比较。<br>2. 空间复杂度：O(1)，原地排序。<br>3. 稳定性：不稳定。例如序列 [5, 8, 5, 2, 9]，第一轮会将第一个 5 和 2 交换，破坏了两个 5 原有的顺序。

```python
# 实现
def selection_sort(arr):
    n = len(arr)
    for i in range(n):
        min_index = i
        for j in range(i+1, n):
            if arr[j] < arr[min_index]:
                min_index = j
        arr[i], arr[min_index] = arr[min_index], arr[i]
    return arr

test_data = [64, 25, 12, 22, 11]
print("原始数组:", test_data)
print("选择排序后:", selection_sort(test_data.copy()))              
```

##### ***插入排序***
###### **核心思想**：通过构建有序序列，对于未排序数据，在已排序序列中从后向前扫描，找到相应位置并插入。这就像我们整理扑克牌一样。<br>**算法步骤**：<br>1. 将第一个元素视为已排序序列。<br>2. 取出下一个元素，在已经排序的元素序列中从后向前扫描。<br>3. 如果该元素（已排序）大于新元素，将该元素移到下一位置。<br>4. 重复步骤 3，直到找到已排序的元素小于或等于新元素的位置。<br>5. 将新元素插入到该位置后。<br>6. 重复步骤 2~5，直到所有元素处理完毕。
###### **性能分析**：<br>1. 时间复杂度：最好情况 O(n)（已有序），最坏和平均情况 O(n²)。<br>2. 空间复杂度：O(1)，原地排序。<br>3. 稳定性：稳定。<br>4. 特点：对于小规模或基本有序的数据，插入排序非常高效。它是高级排序算法（如 Timsort）在小规模数据时切换使用的算法。

```python
# 实现
def insertion_sort(arr):
    n = len(arr)
    for i in range(1, n):
        key = arr[i] # 当前待插入的元素
        j = i-1 # 已排序序列的最后一个索引
        while j >= 0 and key < arr[j]: # 只有key比当前元素小才执行交换操作，否则直接追加。
            arr[j+1] = arr[j]
            j -= 1
        arr[j+1] = key
    return arr

test_data = [12, 11, 13, 5, 6]
print("原始数组:", test_data)
print("插入排序后:", insertion_sort(test_data.copy()))
```
