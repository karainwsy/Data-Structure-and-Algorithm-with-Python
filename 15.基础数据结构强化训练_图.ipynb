##### ***图***
###### 图的主要操作有：添加顶点，添加边，删除边，删除顶点，判断顶点，判断边，获取邻居，深度优先遍历，广度优先遍历

```python
from collections import deque
# 图的定义(用邻接表实现无向图)
class Graph:
    def __int__(self):
        self.adjacency_list = {}
    
    # 添加顶点
    def add_vertex(self, vertex):
        if vertex in self.adjacency_list:
            return 
        self.adjacency_list[vertex] = []
        return 

    # 添加边
    def add_edge(self, vertex1, vertex2):
        # 如果有一个顶点不在就无法添加
        if vertex1 not in self.adjacency_list or vertex2 not in self.adjacency_list:
            return 
        # 如果有重复的就返回
        if vertex2 in self.adjacency_list[vertex1] or vertex1 in self.adjacency_list[vertex2]:
            return 
        self.adjacency_list[vertex1].append(vertex2)
        self.adjacency_list[vertex2].append(vertex1)
        return 

    # 删除边
    def remove_edge(self, vertex1, vertex2):
        # 如果顶点不存在就返回
        if vertex1 not in self.adjacency_list or vertex2 not in self.adjacency_list:
            return 
        # 如果顶点存在但是边不存在，也返回
        if vertex1 not in self.adjacency_list[vertex2]:
            return
        self.adjacency_list[vertex2].remove(vertex1)
        self.adjacency_list[vertex1].remove(vertex2)
        return 

    # 删除顶点
    def remove_vertex(self, vertex):
        if vertex not in self.adjacency_list:
            return 
        for neighbor in self.adjacency_list[vertex]:
            self.adjacency_list[neighbor].remove(vertex)
        del self.adjacency_list[vertex]
        return 

    # 检查顶点是否存在
    def has_vertex(self, vertex):
        return vertex in self.adjacency_list

    # 检查两个顶点之间是否有边
    def has_edge(self, vertex1, vertex2):
        if vertex1 not in self.adjacency_list or vertex2 not in self.adjacency_list:
            return False
        return vertex2 in self.adjacency_list[vertex1]

    # 获取某个顶点的所有邻居
    def get_neighbors(self, vertex):
        if vertex not in self.adjacency_list:
            return None
        return self.adjacency_list[vertex]

    # 深度优先遍历
    def DFS(self, vertex):
        if vertex not in self.adjacency_list:
            return False
        visited = set()
        self._dfs_recursive(vertex, visited)
    def _dfs_recursive(self, vertex, visited):
        visited.add(vertex)
        print(vertex, end=' ')

        neighbor = self.get_neighbors(vertex)
        for nei in neighbor:
            if nei not in visited:
                self._dfs_recursive(nei, visited)
            

    # 广度优先遍历
    def BFS(self, vertex):
        if vertex not in self.adjacency_list:
            return False
        queue = deque()
        visited = set()
        visited.add(vertex)
        queue.append(vertex)
        while queue:
            current_vertex = queue.popleft()
            print(current_vertex, end=' ')
            for neighbor in self.adjacency_list[current_vertex]:
                if neighbor not in visited:
                    queue.append(neighbor)
                    visited.add(neighbor)
```
