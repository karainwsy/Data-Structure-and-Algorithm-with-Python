##### ***栈***
###### 栈的主要操作有：栈类定义、入栈、出栈、查看栈顶、判断栈是否为空、获取栈的大小

```python
# 栈类定义
class Stack:
    def __init__(self):
        self.items = []

    # 入栈
    def push(self, data):
        self.items.append(data)

    # 出栈
    def pop(self):
        return self.items.pop()
    
    # 查看栈顶元素
    def peek(self):
        return self.items[-1]

    # 判断栈是否为空
    def is_empty(self):
        if self.items:
            return False
        return True
        # 或者简化版
        # return not self.items
    
    # 获取栈大小
    def size(self):
        length = 0
        for i in self.items:
            length += 1
        return length

# 测试
test_stack = Stack()

test_stack.push(1)
test_stack.push(2)
print(test_stack.items)       # [1, 2]

print(test_stack.pop())       # 2
print(test_stack.items)       # [1]

print(test_stack.peek())      # 1
print(test_stack.is_empty())  # False
print(test_stack.size())      # 1

print(test_stack.pop())       # 1
print(test_stack.is_empty())  # True
print(test_stack.size())      # 0
```
