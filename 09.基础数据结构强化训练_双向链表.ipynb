##### ***双向链表***
###### 双向链表的操作主要有节点类定义、链表类定义、尾部添加、头部添加、前向遍历、后向遍历、查找、删除、指定位置插入、反转链表

```python
# 节点类定义
class DoublyListNode:
    def __init__(self, data):
        self.data = data
        self.prev = None
        self.next = None

# 链表类定义
class DoublyLinkedList:
    def __init__(self):
        self.head = None

    # 尾部添加
    def append(self, data):
        # 首先创建一个新节点
        new_node = DoublyListNode(data)
        # 判断特殊情况
        if self.head is None:
            self.head = new_node # 空链表没有self.head.next，所以直接插入就好
            return 
        # 先确定链表中的最后一个节点
        last_node = self.head
        while last_node.next:
            last_node = last_node.next
        last_node.next = new_node
        new_node.prev = last_node

    # 头部添加
    def prepend(self, data):
        new_node = DoublyListNode(data)
        # 首先判断边界条件
        if self.head is None:
            self.head = new_node
        # 然后只需将新节点指向头节点，接着更新头节点的引用即可
        new_node.next = self.head
        self.head.prev = new_node
        self.head = new_node

    # 前向遍历
    def traverse_forward(self):
        # 也就是向后遍历，和单链表的遍历一样
        current_node = self.head
        while current_node: # 要处理当前节点就用current_node，要找下一个节点的存在就用current_node.next
            print(current_node.data)
            current_node = current_node.next

    # 后向遍历
    def traverse_backward(self):
        # 与前向遍历相反，所以需要找到最后一个节点
        last_node = self.head
        if last_node is None:
            return
        while last_node.next:
            last_node = last_node.next
        # 找到后执行前向遍历
        while last_node:
            print(last_node.data)
            last_node = last_node.prev

    # 查找
    def search(self, item):
        current_node = self.head
        while current_node:
            if current_node.data == item:
                return current_node
            current_node = current_node.next
        return None

    # 删除
    def delete(self, item):
        current_node = self.head
        # 处理空链表
        if self.head is None:
            return False
        while current_node:
            if current_node.data == item:
                # 单独处理头节点
                if current_node.prev is None:
                    current_node.next.prev = None
                    # 注意更新self.head
                    self.head = current_node.next
                    return True
                # 单独处理尾节点
                if current_node.next is None:
                    current_node.prev.next = None
                    return True
                current_node.next.prev = current_node.prev
                current_node.prev.next = current_node.next
                return True
            current_node = current_node.next
        return False

    # 指定位置插入
    def insert(self, index, data):
        new_node = DoublyListNode(data)
        current_node = self.head
        if index < 0:
            return False
        # 如果插入的是头节点：
        if index == 0:
            new_node.next = current_node
            # 如果非空
            if self.head:
                current_node.prev = new_node
            self.head = new_node
            return True
        for i in range(index):
            # 遍历过程中如果出现了节点为空说明索引越界
            if current_node is None:
                return False
            current_node = current_node.next
        # 如果是尾节点：
        if current_node is None:
            # 此时需要找到原来的尾节点
            last_node = self.head
            while last_node.next:
                last_node = last_node.next
            last_node.next = new_node
            new_node.prev = last_node
            return True
        # 需要先修改新节点后修改当前节点，否则会导致新节点的连接关系出错
        new_node.prev = current_node.prev
        new_node.next = current_node

        current_node.prev.next = new_node
        current_node.prev = new_node
        return True

    # 反转链表
    def reverse(self):
        current_node = self.head
        # 遍历的时候反转
        while current_node:
            current_node.prev, current_node.next = current_node.next, current_node.prev
            if current_node.prev is None:
                self.head = current_node
            current_node = current_node.prev
        return True

# ==================== 测试 ====================

linked_list = DoublyLinkedList()

# 1. 空链表
print("=== 空链表 ===")
print(linked_list.search(10))       # False
print(linked_list.delete(10))       # False
print(linked_list.insert(0, 10))    # True

# 此时：
# 10


# 2. 尾部添加
print("\n=== 尾部添加 ===")
linked_list.append(20)
linked_list.append(30)

linked_list.traverse_forward()
# 10
# 20
# 30

linked_list.traverse_backward()
# 30
# 20
# 10


# 3. 头部添加
print("\n=== 头部添加 ===")
linked_list.prepend(5)

linked_list.traverse_forward()
# 5
# 10
# 20
# 30

linked_list.traverse_backward()
# 30
# 20
# 10
# 5


# 4. 查找
print("\n=== 查找 ===")
print(linked_list.search(20).data)  # 20
print(linked_list.search(100))      # False


# 5. 指定位置插入
print("\n=== 指定位置插入 ===")

print(linked_list.insert(0, 1))     # True
# 1 ⇄ 5 ⇄ 10 ⇄ 20 ⇄ 30

print(linked_list.insert(2, 8))     # True
# 1 ⇄ 5 ⇄ 8 ⇄ 10 ⇄ 20 ⇄ 30

print(linked_list.insert(6, 40))    # True
# 1 ⇄ 5 ⇄ 8 ⇄ 10 ⇄ 20 ⇄ 30 ⇄ 40

print(linked_list.insert(10, 50))   # False


# 6. 删除
print("\n=== 删除 ===")

print(linked_list.delete(1))        # True
# 删除头节点

print(linked_list.delete(40))       # True
# 删除尾节点

print(linked_list.delete(20))       # True
# 删除中间节点

print(linked_list.delete(100))      # False
# 删除不存在的节点

linked_list.traverse_forward()
# 5
# 8
# 10
# 30


# 7. 反向遍历，检查 prev 是否正确
print("\n=== 反向遍历 ===")

linked_list.traverse_backward()
# 30
# 10
# 8
# 5


# 8. 反转
print("\n=== 反转 ===")

linked_list.reverse()

linked_list.traverse_forward()
# 30
# 10
# 8
# 5

linked_list.traverse_backward()
# 5
# 8
# 10
# 30
```
