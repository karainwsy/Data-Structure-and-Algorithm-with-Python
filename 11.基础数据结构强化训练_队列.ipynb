##### ***队列***
###### 队列的主要操作有：队列类定义、入队、出队、查看队头、判断队是否为空、获取队列的大小

```python
# 队列类的定义
class Queue:
    def __init__(self):
        self.items = []
    
    # 入队
    def enqueue(self, data):
        self.items.append(data)
    
    # 出队
    def dequeue(self):
        # 与栈不一样的是，队列遵循先进先出
        return self.items.pop(0)
    
    # 查看队头
    def front(self):
        if self.is_empty():
            return None
        return self.items[0]
    
    # 判断队是否为空
    def is_empty(self):
        return not self.items

    # 获取队列大小
    def size(self):
        return len(self.items)

# 测试
test_queue = Queue()
test_queue.enqueue(1)
test_queue.enqueue(2)
print(test_queue.items)
test_queue.dequeue()
print(test_queue.items)
test_queue.dequeue()
print(test_queue.items)
print(test_queue.front())
print(test_queue.is_empty())
print(test_queue.size())
```
