##### ***数据结构与算法简介***
###### **数据结构**：顾名思义是计算机中数据的存储、组织方式，将物理世界中的数据大致按照一定的逻辑关系分为了几大类数据结构，其目的是让程序能够更加高效地访问和处理数据
###### **算法**：是解决解决特定问题的一系列明确、有限的处理方式。
##### ***常见的数据结构***
###### 按照不同的分类方式，数据结构可分为**逻辑结构**和**物理结构**两大类。
###### **逻辑结构**描述数据元素之间的抽象关系，与计算机如何实现无关。它主要分为：<br> 1. 线性结构：数据元素之间存在"一对一"的线性关系。如数组、链表、栈、队列。<br> 2. 非线性结构： 数据元素之间存在"一对多"或"多对多"的复杂关系。如树、图、集合。
###### **物理结构**（又称存储结构）描述数据在计算机内存中的实际存储方式。它主要分为：<br> 1. 顺序存储： 把逻辑上相邻的元素存储在物理位置也相邻的存储单元中。如数组。<br> 2. 链式存储： 不要求逻辑上相邻的元素在物理位置上也相邻，元素间的逻辑关系通过附加的指针来表示。如链表。

##### ***常用的算法***
###### 算法是解决特定问题的一系列清晰、有限的指令步骤。一个有效的算法应该具备以下五个特性：<br>1. 有穷性： 算法必须在执行有限步之后结束。<br>2. 确定性： 算法的每一步都必须有确切的定义，不会产生二义性。<br>3. 可行性： 算法中的每一步都可以通过已经实现的基本运算执行有限次来实现。<br>4. 输入： 一个算法有零个或多个输入。<br>5. 输出： 一个算法有一个或多个输出，输出是与输入有特定关系的量。
###### 算法的评价标准：时间复杂度与空间复杂度.我们如何判断一个算法的好坏？通常从时间和空间两个维度来衡量。<br>1. 时间复杂度： 定性描述该算法的运行时间。它表示随着问题规模 n 的增大，算法执行时间的增长趋势。常用大 O 符号表示。<br>2. 空间复杂度： 定性描述该算法在运行过程中临时占用存储空间的大小。同样用大 O 符号表示。
###### 算法按用处主要分为：<br>1. 搜索算法：如线性搜索、二分搜索、哈希搜索、深度优先搜索（DFS）、广度优先搜索（BFS）等。<br>2. 排序算法：如冒泡排序、插入排序、选择排序、快速排序、归并排序、基数排序、堆排序等。<br>3. 图算法：如最短路径算法（Dijkstra、Floyd-Warshall、Bellman-Ford）、最小生成树算法（Prim、Kruskal）、拓扑排序、图的遍历算法等。<br>4. 动态规划：用于解决具有重叠子问题和最优子结构性质的问题。<br>5. 分治算法：将问题分解为更小、更简单的子问题，然后递归地解决这些子问题。<br>6. 回溯算法：通过探索所有可能的候选解来找出所有的解。
