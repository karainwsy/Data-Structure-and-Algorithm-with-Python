##### ***二叉树***
###### 二叉树的主要操作有：树节点的定义，节点的插入，节点的四种遍历，搜索，统计节点数量，统计树的层数

```python
from collections import deque
# 树节点的定义
class TreeNode():
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None

# 二叉树的定义
class BinaryTree:
    def __init__(self):
        self.root = None

    # 节点插入
    def insert(self, data):
        # 先创建一个新节点
        new_node = TreeNode(data)
        # 然后先处理异常情况
        if self.root is None:
            self.root = new_node
            return True
        # 非空情况下，使用队列
        queue = deque([self.root]) # deque需要一个可迭代的对象，所以将root作为一个列表元素
        while queue:
            current_node = queue.popleft()
            if current_node.left:
                queue.append(current_node.left)
            else:
                current_node.left = new_node
                return True
            if current_node.right:
                queue.append(current_node.right)
            else:
                current_node.right = new_node
                return True

    # 前序遍历（递归实现）
    def preorder(self, node):
        if node is None:
            return 
        print(node.data, end=' ')
        self.preorder(node.left)
        self.preorder(node.right)

    # 中序遍历（递归实现）
    def inorder(self, node):
        if node is None:
            return 
        self.inorder(node.left)
        print(node.data, end=' ')
        self.inorder(node.right)

    # 后序遍历（递归实现）
    def postorder(self, node):
        if node is None:
            return 
        self.postorder(node.left)
        self.postorder(node.right)
        print(node.data, end=' ')

    # 层序遍历
    def levelorder(self, node):
        if node is None:
            return 
        queue = deque([node])
        while queue:
            current_node = queue.popleft()
            print(current_node.data, end=' ')
            if current_node.left:
                queue.append(current_node.left)
            if current_node.right:
                queue.append(current_node.right)

    # 搜索
    def search(self, data):
        if self.root is None:
            return False
        queue = deque([self.root])
        while queue:
            current_node = queue.popleft()
            if current_node.data == data:
                return True
            if current_node.left:
                queue.append(current_node.left)
            if current_node.right:
                queue.append(current_node.right)
        return False

    # 统计节点数量
    def size(self):
        counter = 0
        if self.root is None:
            return counter
        queue = deque([self.root])
        while queue:
            current_node = queue.popleft()
            counter += 1
            if current_node.left:
                queue.append(current_node.left)
            if current_node.right:
                queue.append(current_node.right)
        return counter

    # 计算树的层数
    def height(self):
        tree_height = 0
        if self.root is None:
            return tree_height
        queue = deque([self.root])
        while queue:
            level_size = len(queue)
            for _ in range(level_size):
                current_node = queue.popleft()
                if current_node.left:
                    queue.append(current_node.left)
                if current_node.right:
                    queue.append(current_node.right)
            tree_height += 1
        return tree_height
```

##### ***二叉搜索树***
###### 与二叉树最大的区别在于它有顺序要求，即左子树大小 < 当前节点 < 右子树，所以对于它的插入和搜索不再需要层序遍历而是可以直接对数值进行判断即可。

```python
class TreeNode:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None
# 二叉搜索树的定义
class BinarySearchTree:
    def __init__(self):
        self.root = None

    # 插入
    def insert(self, data):
        new_node = TreeNode(data)
        if self.root is None:
            self.root = new_node
            return True
        current_node = self.root
        while current_node:
            if data < current_node.data:
                if current_node.left is None:
                    current_node.left = new_node
                    return True
                else:
                    current_node = current_node.left
            elif data > current_node.data:
                if current_node.right is None:
                    current_node.right = new_node
                    return True
                else:
                    current_node = current_node.right
            else: 
                return False

    # 搜索
    def search(self, data):
        current_node = self.root
        while current_node:
            if data == current_node.data:
                return True
            elif data < current_node.data:
                current_node = current_node.left
            else:
                current_node = current_node.right
        return False

    # 搜索最大最小值
    def find_min(self):
        if self.root is None:
            return None
        current_node = self.root
        while current_node.left:
            current_node = current_node.left
        return current_node.data
    def find_max(self):
        if self.root is None:
            return None
        current_node = self.root
        while current_node.right:
            current_node = current_node.right
        return current_node.data

    # 删除，注意删除后依然要保证BST的特性
    def delete(self, data):
        parent_node = None
        current_node = self.root
        while current_node:
            if data < current_node.data:
                if current_node.left is None:
                    return False
                parent_node = current_node
                current_node = current_node.left
            elif data > current_node.data:
                if current_node.right is None:
                    return False
                parent_node = current_node
                current_node = current_node.right
            else:
                # 删除叶子节点的情况
                if current_node.left is None and current_node.right is None:
                    # 如果删除的是根节点
                    if parent_node is None:
                        self.root = None
                        return True
                    # 如果是左孩子
                    if parent_node.left is current_node:
                        parent_node.left = None
                    # 如果是右孩子
                    else: 
                        parent_node.right = None
                    return True

                # 删除只有左孩子的情况
                if current_node.left and current_node.right is None:
                    # 如果删除的是根节点
                    if parent_node is None:
                        self.root = current_node.left
                        return True
                    if parent_node.left is current_node:
                        parent_node.left = current_node.left
                    else:
                        parent_node.right = current_node.left
                    return True
                # 删除只有右孩子的情况
                if current_node.left is None and current_node.right:
                    if parent_node is None:
                        self.root = current_node.right
                        return True
                    if parent_node.left is current_node:
                        parent_node.left = current_node.right
                    else:
                        parent_node.right = current_node.right
                    return True

                # 删除有两个孩子的情况，关键在于找到删除节点的右子树的最小节点
                else:
                    right_tree_parent = current_node # 要删除的节点
                    right_tree_root = current_node.right # 删除节点的右子树根节点
                    # 找到最小的也就是find_min()
                    while right_tree_root.left:
                        right_tree_parent = right_tree_root
                        right_tree_root = right_tree_root.left
                    current_node.data = right_tree_root.data
                    # 如果这个最小值就是当前节点的右孩子
                    if right_tree_parent is current_node:
                        current_node.right = right_tree_root.right # 为什么不直接用None是因为，可能还有右孩子，下面同理
                    # 如果这个最小值还在更深处，则需要考虑处理最小值下面可能存在的右孩子
                    else:
                        right_tree_parent.left = right_tree_root.right

    # 遍历
    def preorder(self, node):
        if node is None:
            return 
        print(node.data, end=' ')
        self.preorder(node.left)
        self.preorder(node.right)
    def inorder(self, node):
        if node is None:
            return 
        self.inorder(node.left)
        print(node.data, end=' ')
        self.inorder(node.right)
    def postorder(self, node):
        if node is None:
            return 
        self.postorder(node.left)
        self.postorder(node.right)
        print(node.data, end=' ')

                
```
