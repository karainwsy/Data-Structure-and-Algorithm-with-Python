##### ***树形结构***
###### 树（Tree）是一种非线性的数据结构，它模拟了自然界中树的分支层次，与数组、链表这种一个接一个的线性结构不同，树中的数据元素（称为节点）之间存在明确的一对多的层次关系。<br>树形结构就像是一棵真实的树：<br>1. 根节点：树的根部，唯一没有父节点的节点<br>2. 子节点：从某个节点分支出去的节点<br>3. 叶节点：没有子节点的节点，像树叶<br>4. 路径：从一个节点到另一个节点的路线<br>5. 深度：从根节点到该节点的路径长度<br>6. 高度：从该节点到最远叶节点的路径长度
<font size=3>

| 操作 | 数组/链表 | 哈希表 | 删除效率 | 有序性 |
| :---: | :---: | :---: | :---: | :---: |
| 数组 | O(n) | O(1) | O(n) | 无 |
| 链表 | O(n) | O(1) | O(1) | 无 |
| 哈希表 | O(1) | O(1) | O(1) | 无 |
| 二叉搜索树 | O($\log n$) | O($\log n$) | O($\log n$) | 有 |</font>
###### 接下来熟悉一下树的基本术语：<br>1. 节点（Node）：树中的每一个元素。它包含存储的数据和指向其子节点的链接。<br>2. 根节点（Root）：位于树顶端的节点，是整棵树的起点。一棵树有且仅有一个根节点。<br>3. 父节点与子节点（Parent & Child）：如果一个节点 A 连接到下方的节点 B，那么 A 是 B 的父节点，B 是 A 的子节点。一个父节点可以有多个子节点。<br>4. 兄弟节点（Siblings）：拥有同一个父节点的节点们互为兄弟节点。<br>5. 叶子节点（Leaf）：没有子节点的节点，也称为终端节点。<br>6. 边（Edge）：连接两个节点的线，表示它们之间的关系。<br>7. 路径（Path）：从根节点到某个特定节点所经过的节点序列。<br>8. 高度（Height）：从某个节点到其最远叶子节点的最长路径上的边数。树的高度即根节点的高度。<br>9. 深度（Depth）：从根节点到某个节点的路径上的边数。根节点的深度为 0。<br>10. 层（Level）：深度相同的所有节点属于同一层。根节点在第 0 层。

###### 树的基本性质如下：<br>1. 唯一路径：树中任意两个节点之间有且仅有一条路径。<br>2. N 个节点，N-1 条边：一棵具有 N 个节点的树，总共有 N-1 条边。<br>3. 无环：树中不存在环（即从某个节点出发，沿着边最终又能回到该节点）。

###### 作为同一家族的数据结构，它们的构造其实大差不差，在此介绍简单的普通二叉树及其相关操作：查找、插入、删除、遍历。

```python
# 利用Python代码实现二叉树
class TreeNode:
    def __init__(self, value):
        self.value = value
        self.left = None
        self.right = None

    def __str__(self):
        return f"Node({self.value})"
```

```python
# 创建节点
root = TreeNode(1)
node2 = TreeNode(2)
node3 = TreeNode(3)
node4 = TreeNode(4)
node5 = TreeNode(5)
node6 = TreeNode(6)

# 构建连接关系
root.left = node2
root.right = node3
node2.left = node4
node2.right = node5
node3.right = node6
print(f"根节点: {root}")
print(f"根节点的左子节点: {root.left}")
print(f"根节点的右子节点: {root.right}")
```

###### 构建好一颗二叉树后，我们来实现它的操作，而所有的操作都需要先知道数据的位置，所以我们来实现遍历操作，其中关于二叉树有四种遍历算法：<br>1. 前序遍历：先访问根节点，然后递归地前序遍历左子树，然后递归地遍历右子树。<br>2. 中序遍历：先递归地中序遍历左子树，然后访问根节点，最后递归地中序遍历右子树<br>3. 后序遍历：先递归地后序遍历左子树，然后递归地后序遍历右子树，最后访问根节点。<br>4. 层序遍历：从根节点开始，一层一层地访问节点，从左到右。
        1
       / \
      2   3
     / \   \
    4   5   6
###### 前序：1, 2, 4, 5, 3, 6<br>中序：4, 2, 5, 1, 3, 6<br>后序：4, 5, 2, 6, 3, 1<br>层序：1, 2, 3, 4, 5, 6

```python
# 使用递归实现树遍历
def preorder_traversal(node):
    # 前序遍历的顺序是根 -> 左 -> 右
    if node is None:
        return 
    print(node.value, end=' ')
    preorder_traversal(node.left)
    preorder_traversal(node.right)

def inorder_traversal(node):
    # 中序遍历的顺序是左 -> 根 -> 右
    if node is None:
        return
    inorder_traversal(node.left)
    print(node.value, end=' ')
    inorder_traversal(node.right)

def postorder_traversal(node):
    """后序遍历：左 -> 右 -> 根"""
    if node is None:
        return
    postorder_traversal(node.left)
    postorder_traversal(node.right)
    print(node.value, end=' ')  

print("\n--- 遍历演示 ---")
print("前序遍历结果:", end=' ')
preorder_traversal(root) # 输出: 1 2 4 5 3 6

print("\n中序遍历结果:", end=' ')
inorder_traversal(root)  # 输出: 4 2 5 1 3 6

print("\n后序遍历结果:", end=' ')
postorder_traversal(root) # 输出: 4 5 2 6 3 1
```

```python
from collections import deque
# 实现层序遍历
def level_order_traversal(root):
    # 层序遍历：使用队列，将每一层的节点放到队列中，然后按从左到右出队
    if root is None:
        return

    queue = deque([root]) # 初始化队列，放入根节点

    while queue:
        current_node = queue.popleft()
        print(current_node.value, end=' ')

        # 然后将每一层的下一层的节点添加到队列中
        if current_node.left:
            queue.append(current_node.left)
        if current_node.right:
            queue.append(current_node.right)
print("\n层序遍历结果:", end=' ')
level_order_traversal(root) 
```
