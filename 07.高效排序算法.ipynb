##### ***$O(n\log n )$级高效排序算法***
###### 当数据量变大时，O(n²) 的算法会变得非常慢。下面这些更高效的算法是工程实践中的主力。

##### ***快速排序***
###### **核心思想**：分治法。选择一个基准元素，通过一趟排序将待排记录分隔成独立的两部分，其中一部分记录的关键字均比另一部分的关键字小，然后分别对这两部分记录继续进行排序，以达到整个序列有序。<br>**算法步骤**：<br>1. 选择基准：从数列中挑出一个元素，称为"基准"。<br>2. 分区操作：重新排列数列，所有比基准值小的元素摆放在基准前面，所有比基准值大的元素摆在基准后面（相同的数可以到任一边）。在这个分区退出之后，该基准就处于数列的中间位置。这个称为分区操作。<br>3. 递归排序：递归地将小于基准值元素的子数列和大于基准值元素的子数列排序。
###### **性能分析**：<br>1. 时间复杂度：平均情况 O(n log n)，最坏情况 O(n²)（当数组已有序或逆序，且基准选择不当时）。通过随机选择基准或"三数取中"法可以极大避免最坏情况。<br>2. 空间复杂度：平均 O(log n)（递归调用栈），最坏 O(n)。<br>3. 稳定性：不稳定。

```python
# 实现
def quick_sort(arr):
    # 由于该算法的特殊性，需要额外编写一个分区函数
    def _quick_sort(arr, low, high):
        if low < high:
            pi = partition(arr, low, high)
            # 递归的处理分区后的子区域
            _quick_sort(arr, low, pi - 1)
            _quick_sort(arr, pi + 1, high)
    
    def partition(arr, low, high):
        pivot = arr[high] # 首先选取arr的末尾当做基准
        i = low - 1 # 设置一个标记记录小于基准的序列的最后一个索引用于后续交换
        for j in range(low, high):
            # 如果当前元素小于或等于基准
            if arr[j] <= pivot:
                i += 1 # 此时说明已经有一个元素需要放在小于的区域了，所以同步将标记加1
                arr[i], arr[j] = arr[j], arr[i] # 然后将已有区域的后一个和这个小于基准的值互换
        # 循环结束后单独处理pivot 
        arr[i+1], arr[high] = arr[high], arr[i+1]
        return i + 1

    _quick_sort(arr, 0, len(arr)-1)
    return arr
    
test_data = [10, 80, 30, 90, 40, 50, 70]
print("原始数组:", test_data)
print("快速排序后:", quick_sort(test_data.copy()))

```

##### ***归并排序***
###### **核心思想**：分治法的典型应用。将已有序的子序列合并，得到完全有序的序列。即先使每个子序列有序，再使子序列段间有序。<br>**算法步骤**：<br>1. 分解：将当前区间一分为二，即求分裂点 mid = (low + high)/2。<br>2. 求解：递归地对两个子区间 arr[low...mid] 和 arr[mid+1...high] 进行归并排序。递归的终止条件是子区间长度为1（认为一个元素自然有序）。<br>3. 合并：将两个已排序的子区间合并为一个有序区间。
###### **性能分析**：<br>1. 时间复杂度：最好、最坏、平均情况均为 O(n log n)。性能非常稳定。<br>2. 空间复杂度：O(n)，需要与原始数组等大的额外空间来合并。<br>3. 稳定性：稳定。


```python
def merge_sort(arr):
    n = len(arr)
    if n > 1:
        mid = n // 2 # 找到中间点，进行分割
        left_half = arr[:mid]
        right_half = arr[mid:]

        # 递归调用左右两版进行排序
        merge_sort(left_half)
        merge_sort(right_half)

        # 合并两个已排序的子数组
        i = j = k = 0 # 分别为左右字数组和原数组创建标记
        while i < len(left_half) and j < len(right_half):
            if left_half[i] < right_half[j]:
                arr[k] = left_half[i] # 添加小的
                i += 1
            else:
                arr[k] = right_half[j]
                j += 1
            k += 1 # 每添加一个，索引便向后移
        # 检查左右子数组是否有剩余元素
        while i < len(left_half):
            arr[k] = left_half[i]
            i += 1
            k += 1
        while j < len(right_half):
            arr[k] = right_half[j]
            j += 1
            k += 1
    return arr

test_data = [38, 27, 43, 3, 9, 82, 10]
print("原始数组:", test_data)
print("归并排序后:", merge_sort(test_data.copy()))
```
