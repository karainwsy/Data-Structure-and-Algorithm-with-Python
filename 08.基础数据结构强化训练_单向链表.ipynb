##### ***单向链表***
###### 单向链表的操作主要有节点类定义、链表类定义、尾部添加、头部添加、遍历、查找、删除、根据下标获取节点、指定位置插入、获取长度、反转链表

```python
# 节点类定义
class ListNode:
    def __init__(self, data):
        self.data = data # 数据域
        self.next = None # 指针域

# 链表类定义
class LinkedList:
    def __init__(self): # 初始化一个空链表，并让head节点指向None
        self.head = None
    
    # 尾部添加
    def append(self, data):
        # 添加之前首先需要创建一个新节点
        new_node = ListNode(data)
        # 然后判断原本的链表是否为空
        if self.head is None:
            self.head = new_node
            return 
        # 如果不为空就要把新节点添加到尾部，所以需要先找到尾节点
        last_node = self.head # 先假定为尾节点就是头节点，然后while循环找真正的尾节点（指向None）
        while last_node.next:
            last_node = last_node.next
        last_node.next = new_node

    # 头部添加
    def prepend(self, data):
        # 头部添加也需要先创建一个新节点
        new_node = ListNode(data)
        # 然后由于head节点的存在，所以可以直接操作head节点进行连接
        if self.head is None:
            self.head = new_node
            return
        temp_node = self.head
        new_node.next = temp_node
        self.head = new_node
        # 简洁法：由于是空链表，所以本质上头部添加逻辑是一致的可以省略此判断，而后续的非空链表可以先让新节点接到head，然后让head引用回新节点
        """
        def prepend(self, data):
            new_node = ListNode(data)
            new_node.next = self.head
            self.head = new_node
        """
    
    # 遍历
    def traverse(self):
        # 先确定遍历的开始点，head节点
        current_node = self.head
        # 然后只要current不为空就一直向后遍历
        while current_node:
            print(current_node.data)
            current_node = current_node.next

    # 查找
    def search(self, item):
        # 查找本质是遍历，只是多了一步数值匹配
        current_node = self.head
        while current_node:
            if current_node.data == item:
                return True
            current_node = current_node.next
        return False

    # 删除第一个符合的节点
    def delete(self, item):
        # 删除也算遍历的一种变体，所以需要一个节点来标记遍历到哪了，此外如果删除中间节点，还需额外维护一个当前节点的前一个节点
        current_node = self.head
        pre_node = None
        # 但是删除需要考虑多种情况，首先考虑表为空的情况
        if self.head is None:
            return False
        # 其次考虑删除头节点
        if self.head.data == item:
            self.head = self.head.next
            return True
        # 接着考虑删除中间节点
        while current_node:
            if current_node.data == item:
                # 找到后需要将该节点的前一个节点链接到后一个节点
                pre_node.next = current_node.next
                return True
            pre_node = current_node
            current_node = current_node.next
        return False

    # 根据下标获取节点
    def get(self, index):
        current_node = self.head
        # 根据下标获取节点，也就是有次数的循环查找
        if index < 0:
            return None
        for _ in range(index):
            current_node = current_node.next
            if current_node is None:
                return None
        return current_node

    # 指定位置插入
    def insert(self, index, data):
        # 插入和删除类似，需要操作当前节点和上一节点
        current_node = self.head
        pre_node = None
        new_node = ListNode(data)
        # 首先处理异常情况
        if index < 0:
            return False
        # 接着考虑index == 0也就是插入头结点，退化成头部添加
        if index == 0:
            new_node.next = self.head
            self.head = new_node
            return True
        # 然后是在中间插入
        for i in range(index):
            if current_node is None: # 如果在末尾插入，本质上可以不用做判断，但是如果索引超过了边界，则需要提前终止
                return False
            pre_node = current_node
            current_node = current_node.next
        pre_node.next = new_node
        new_node.next = current_node
        return True

    # 获取长度
    def length(self):
        length = 0
        current_node = self.head
        while current_node:
            length += 1
            current_node = current_node.next
        return length

    # 反转链表
    def reverse(self):
        # 链表的反转比较复杂，需要同时控制，前一个，当前，后一个节点的连接关系，所以需要维护三个节点
        current_node = self.head
        pre_node = None
        # 接着操作当前节点向后遍历，同时修改连接关系
        while current_node:
            # 首先由于遍历和反转操作会导致current节点的next丢失，所以需要next_node来保存
            next_node = current_node.next
            # 接着进行反转操作
            current_node.next = pre_node
            # 而后为了后续的遍历需要pre_node跟上current_node
            pre_node = current_node
            # 最后更新current_node去next节点
            current_node = next_node
        # 遍历结束后需要更新头节点的位置
        self.head = pre_node
        return True

# 测试
linked_list = LinkedList()

linked_list.append(10)
linked_list.append(20)
linked_list.append(30)

linked_list.traverse()
# 10 → 20 → 30

linked_list.prepend(5)
linked_list.traverse()
# 5 → 10 → 20 → 30

linked_list.insert(2, 15)
linked_list.traverse()
# 5 → 10 → 15 → 20 → 30

print(linked_list.search(20))
# True

print(linked_list.length())
# 5

node = linked_list.get(2)
print(node.data)
# 15

linked_list.delete(15)
linked_list.traverse()
# 5 → 10 → 20 → 30

linked_list.reverse()
linked_list.traverse()
# 30 → 20 → 10 → 5
```
